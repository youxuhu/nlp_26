```python
text = 'You say goodbye and I say hello.'
```

```python
text = text.lower()
text
```

```python
text = text.replace('.', ' .')
words = text.split(' ')
text
```

```python
words
```

```python
word_to_id = {}
id_to_word = {}
for word in words:
    if word not in word_to_id:
        new_id = len(word_to_id)
        word_to_id[word] = new_id
        id_to_word[new_id] = word
id_to_word
word_to_id
```

```python
import numpy as np
corpus = [word_to_id[w] for w in words]
corpus = np.array(corpus)
corpus
```
